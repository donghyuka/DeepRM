```python
df["exonStarts"] = df["exonStarts"].apply(lambda x: x.split(",")).apply(lambda x: [int(i) for i in x])
df["exonEnds"] = df["exonEnds"].apply(lambda x: x.split(",")).apply(lambda x: [int(i) for i in x])
```

```python
print(df)
```

```python
df["txStart"] = df["txStart"].astype(int) -1
df["txEnd"] = df["txEnd"].astype(int)
df["cdsStart"] = df["cdsStart"].astype(int) -1
df["cdsEnd"] = df["cdsEnd"].astype(int)
```

```python
df["exonStarts"] = df["exonStarts"].apply(lambda x: [i-1 for i in x])
```

```python
df["cdsStart"] = df.apply(lambda x: x["cdsStart"] if x["coding"] else x["cdsEnd"], axis=1)
```

```python
print(df)
```

```python
df["exonStarts"] = df["exonStarts"].apply(lambda x: ",".join([str(i) for i in x])+ ",")
df["exonEnds"] = df["exonEnds"].apply(lambda x: ",".join([str(i) for i in x])+ ",")
```
