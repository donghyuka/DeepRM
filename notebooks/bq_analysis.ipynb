```python
## Plot inverse CDF
fig, ax = plt.subplots(figsize=(10, 10))


ax.plot(np.linspace(0, 1, len(ca_df), endpoint=False),  np.log10(np.sort(ca_df.values)[::-1]), label="A", color="royalblue")
ax.plot(np.linspace(0, 1, len(m6a_df), endpoint=False), np.log10(np.sort(m6a_df.values)[::-1]), label="m6A", color="tomato")
ax.set_xlabel("Fraction")
ax.set_ylabel("log10(Count)")
ax.set_ylim(0, 4.5)
ax.legend()
plt.savefig("/extdata4/baeklab/Hyeonseo/m6A/poster/motif_cdf.pdf")
plt.show()

```
