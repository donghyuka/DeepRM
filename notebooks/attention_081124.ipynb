```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import os
import seaborn as sns
import sys
sys.path.append("/extdata4/baeklab/Hyeonseo/m6A/modformer")
from utils.utils import printmessage
import glob
import tqdm
```

```python
sample = "/extdata4/baeklab/Hyeonseo/m6A/activation/inference/AIRNA-DW-v4-20240905-092858-10-177000-ON0099/inference_0_0.npz"
with np.load(sample) as data:
    data = {key: data[key] for key in data}
```

```python
print(data.keys())
```

```python
print(data["label_id"].shape)
print(data["pred"].shape)
print(data["activation"].shape)
```

```python
## Visualize the activation
activation = data["activation"].mean(axis=0)
fig, ax = plt.subplots(2, 3, figsize=(20, 15))
cmap = sns.diverging_palette(130, 10, s=90, l=60, n=9, center="dark", as_cmap=True)
for i in range(6):
    sns.heatmap(activation[i].T, ax=ax[i//3, i%3], cmap="icefire", center=0)
    head_boundaries = activation[i].shape[-1]//8 * np.arange(1, 8)
    for boundary in head_boundaries:
        ax[i//3, i%3].axhline(boundary, color="w")
    ax[i//3, i%3].set_title(f"Layer {i+1}")
plt.show()
```
