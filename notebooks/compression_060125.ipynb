```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import os
import seaborn as sns
import sys
import glob
import tqdm
import pickle
plt.style.use('default')
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams.update({'font.size': 28, 'legend.facecolor': 'white', 'legend.framealpha': 1, "legend.frameon": 1, "lines.linewidth": 2})
```

```python
path = "/extdata4/baeklab/Hyeonseo/m6A/dataset/ver112524_exc9398/score-1.0/val/pos"
```

```python
paths = glob.glob(f"{path}/*.npz")
```

```python
with np.load(paths[0]) as data:
    key = data.files
print(key)
```

```python
data_dict = {k: [] for k in key}
for p in tqdm.tqdm(paths):
    with np.load(p) as data:
        for k in key:
            data_dict[k].append(data[k])
```

```python
for k in key:
    data_dict[k] = np.concatenate(data_dict[k])
    
```

```python
outdir = "/extdata4/baeklab/Hyeonseo/m6A/dataset/comp_test"
os.makedirs(outdir, exist_ok=True)
```

```python
np.savez_compressed(f"{outdir}/data.compressed.npz", **data_dict)
np.savez(f"{outdir}/data.npz", **data_dict)
```

```python
df = pd.DataFrame(data_dict)
df.to_pickle(f"{outdir}/data.pkl")
```
