```python
## Groupby 5mer and calculate mean signal
def get_motif_signal(df_dict, key = "5mer"):
    motif_dict = {}
    for k in df_dict.keys():
        df = df_dict[k]
        motif_df = df[["bq_centre", "signal_centre", "bq_avg", "signal_avg", key]].groupby(key).mean()
        motif_count_df = df[key].value_counts()
        motif_df = pd.merge(motif_df, motif_count_df, left_index = True, right_index = True)
        motif_dict[k] = motif_df
    return motif_dict
```

```python
## Plot scatter : ON0092 vs. ON0096

from scipy import stats

def get_density(x,y, sample=1000, bw_method= 0.2):
    values = np.stack([x, y], axis=0)
    sampled = values
    sampled = sampled[:, np.random.choice(sampled.shape[1], sample, replace=False)]
    kernel = stats.gaussian_kde(sampled, bw_method=bw_method)
    density = kernel(values)
    density = np.clip(density, None, np.percentile(density[x>0], 99))
    return density

def plot(ax, df_dict, k1, k2, feature = "signal_avg", x_label = None, y_label = None, min_count = 10):
    merged_df = pd.merge(df_dict[k1][df_dict[k1]["count"] > min_count][[feature]],
                         df_dict[k2][df_dict[k2]["count"] > min_count][[feature]],
                         how = "inner", left_index = True, right_index = True).dropna()
    merged_df = merged_df[[f"{feature}_x", f"{feature}_y"]]
    merged_df.columns = [k1, k2]
    density = get_density(merged_df[k1], merged_df[k2])
    sns.scatterplot(x = k1, y = k2, data = merged_df, ax = ax, c=density,
    cmap=sns.color_palette("flare_r", as_cmap=True), legend=False, linewidth=0, alpha = 1)
    r2 = stats.pearsonr(merged_df[k1], merged_df[k2])[0]**2
    rho2 = stats.spearmanr(merged_df[k1], merged_df[k2])[0]**2
    ax.text(0.03, 0.8, f"R2: {r2:.2f}\nRho2: {rho2:.2f}", transform=ax.transAxes, fontsize=22)
    ax.set_title(feature)
    min_val = min(np.min(merged_df[k1]), np.min(merged_df[k2]))
    max_val = max(np.max(merged_df[k1]), np.max(merged_df[k2]))
    ax.set_xlim(min_val, max_val)
    ax.set_ylim(min_val, max_val)
    ax.set_aspect('equal', 'box')
    ax.set_xlabel(x_label if x_label else k1)
    ax.set_ylabel(y_label if y_label else k2)
    ## plot trendline
    sample_df = merged_df.sample(n=1000)
    z = np.polyfit(sample_df[k1]*100, sample_df[k2]*100, 1)
    x = np.linspace(min_val*100, max_val*100, 100)
    y = np.polyval(z, x)
    x = x/100
    y = y/100
    ax.plot(x,y,linestyle="--", color = "aqua")
    
```

```python
def trim_df(df, feature, ratio = 0.001):
    min_val = df[feature].quantile(ratio)
    max_val = df[feature].quantile(1 - ratio)
    df = df[(df[feature] > min_val) & (df[feature] < max_val)]
    return df
```

```python
def plot_distribution(ax, a1, a2):
    sns.histplot(a1, ax = ax, color = "royalblue", bins = 50, stat="density", fill=False,
                 linewidth = 3, element="step", common_norm = False)
    sns.histplot(a2, ax = ax, color = "tomato", bins = 50, stat="density", fill=False,
                 linewidth = 3, element="step", common_norm = False)
    ## Vline at the mean
    ax.axvline(a1.mean(), color = "royalblue", ls = "--", lw = 2)
    ax.axvline(a2.mean(), color = "tomato", ls = "--", lw = 2)
```
