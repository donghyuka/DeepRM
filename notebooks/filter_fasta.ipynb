```python
import gc
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import os
import seaborn as sns
import sys
import glob
import tqdm
import gc
from Bio import SeqIO
```

```python
fasta_1_path = "/extdata4/baeklab/Hyeonseo/m6A/res/ref/GCF_000001405.40_GRCh38.p14_rna.fna"
fasta_2_path = "/extdata4/baeklab/Hyeonseo/m6A/res/ref/Homo_sapiens.GRCh38.cdna.all.fa"
fasta_3_path = "/extdata4/baeklab/Hyeonseo/m6A/res/ref/Homo_sapiens.GRCh38.ncrna.fa"
refflat_path = "/extdata4/baeklab/Hyeonseo/m6A/anno/agat_refflat.base0.pkl"
```

```python
## Load the FASTA files.
```

```python
fasta_1 = SeqIO.parse(fasta_1_path, format="fasta")
fasta_2 = SeqIO.parse(fasta_2_path, format="fasta")
fasta_3 = SeqIO.parse(fasta_3_path, format="fasta")
```

```python
fasta_1 = SeqIO.to_dict(fasta_1)
fasta_2 = SeqIO.to_dict(fasta_2)
fasta_3 = SeqIO.to_dict(fasta_3)
```

```python
fasta_dict = {}
fasta_dict.update(fasta_1)
fasta_dict.update(fasta_2)
fasta_dict.update(fasta_3)
```

```python
del fasta_1, fasta_2, fasta_3
gc.collect()
```

```python
print(len(fasta_dict))
```

```python
refflat_df = pd.read_pickle(refflat_path)
print(refflat_df)
```

```python
transcript_id = refflat_df["transcript_id"].unique()
```

```python
## Remove trailing decimal in ENSEMBL

new_fasta_dict = {}
for k, v in fasta_dict.items():
    if k.startswith("ENST"):
        k = k.split(".")[0]
    new_fasta_dict[k]=v
```

```python
filtered_fasta = {}
unassigned = []
for k in transcript_id:
    if k in new_fasta_dict:
        filtered_fasta[k] = new_fasta_dict[k]
    else:
        unassigned.append(k)
        
```
