```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import os
import seaborn as sns
import glob
import tqdm
from scipy import stats
```

```python
# Load data
data = pd.read_csv("/extdata4/baeklab/Hyeonseo/m6A/aqseq/dimers.rep56.seq12.2-2.tsv", sep="\t") 

for rep in ['rep1', 'rep2']:
    sum_reads = data[f'nreads_{rep}'].sum()
    ratio = 1 / (4 ** 4)
    ## Normal approximation to the hypergeometric distribution. 
    normal_approx = stats.norm(sum_reads * ratio, np.sqrt(sum_reads * ratio * (1 - ratio)).astype(np.float64))
    ## Apply Continuity correction
    p = 1 - normal_approx.cdf(data[f'nreads_{rep}'] - 0.5)
    data[f'p_{rep}'] = p
    ## Spply Benjamini-Hochberg procedure
    data = data.sort_values(f'p_{rep}', ascending=True)
    data["rank_temp"] = range(1, len(data) + 1)
    data[f'q_{rep}'] = data[f'p_{rep}'] * len(data) / data["rank_temp"]
    data = data.drop(columns=["rank_temp"])
    data[f"fold_{rep}"] = data[f'nreads_{rep}'] / (sum_reads * ratio)

data = data.sort_values("percent_avg", ascending=False)
print(data[(data["q_rep1"] < 0.05) & (data["q_rep2"] < 0.05)])
data.to_csv("/extdata4/baeklab/Hyeonseo/m6A/aqseq/dimers.rep56.seq12.2-2.qvalue.tsv", sep="\t", index=False)
    
    
```
